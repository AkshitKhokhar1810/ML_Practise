```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import sklearn.linear_model
import os
```

```python
# Option 2: Change to script directory (if notebook differs)
os.chdir(r'C:\Users\khokh\Jupyter_Codes\Machine_learning',)  # Replace with actual path
oecd_bli = pd.read_csv('BetterLifeIndex2015.csv')
gdp_per_capita=pd.read_csv('gdp_per_capita.csv',thousands=',',delimiter='\t',encoding='latin1',na_values='n/a')
```

```python
# prepare the data 
country_stats=pd.merge(oecd_bli, gdp_per_capita, on='Country')
X=np.c_[country_stats["Subject Descriptor"]]
Y=np.c_[country_stats["  Life satisfaction"]]
```

```python
country_stats.plot(kind='scatter',x='Subject Descriptor',y='  Life satisfaction')
plt.show()
```

```python
model = sklearn.linear_model.LinearRegression()
```

```python
model.fit(X,Y)
```
